### Basic Arithmetic Operations

Let's explore common arithmetic operations in Python: addition, subtraction, multiplication, and division.

```python
# Addition
num1 = 10
num2 = 5
sum_result = num1 + num2
print(f"The sum of {num1} and {num2} is: {sum_result}")
```

```python
# Subtraction
num1 = 15
num2 = 7
difference_result = num1 - num2
print(f"The difference between {num1} and {num2} is: {difference_result}")
```

```python
# Multiplication
num1 = 6
num2 = 4
product_result = num1 * num2
print(f"The product of {num1} and {num2} is: {product_result}")
```

```python
# Division
num1 = 20
num2 = 5
quotient_result = num1 / num2
print(f"The quotient of {num1} divided by {num2} is: {quotient_result}")

# Integer Division (returns only the integer part)
num3 = 21
num4 = 5
integer_quotient = num3 // num4
print(f"The integer quotient of {num3} divided by {num4} is: {integer_quotient}")
```

```python
# Modulus (returns the remainder of a division)
num5 = 22
num6 = 5
remainder = num5 % num6
print(f"The remainder of {num5} divided by {num6} is: {remainder}")
```

```python
# Exponentiation (raises a number to a power)
base = 2
exponent = 3
power_result = base ** exponent
print(f"{base} raised to the power of {exponent} is: {power_result}")
```

These are the fundamental arithmetic operations. Feel free to modify the numbers and experiment with different values! What would you like to try next?

### Conditional Statements (if/else)

Conditional statements allow your program to make decisions based on certain conditions. Let's look at `if`, `if-else`, and `if-elif-else`.

```python
# Basic 'if' statement
# This code will only execute if the condition is True.

number = 10

if number > 0:
    print(f"{number} is a positive number.")

number = -5
if number > 0:
    print(f"{number} is a positive number.") # This line won't be printed
```

```python
# 'if-else' statement
# This allows you to execute one block of code if the condition is True, and another if it's False.

age = 18

if age >= 18:
    print("You are old enough to vote.")
else:
    print("You are not old enough to vote yet.")

age = 16

if age >= 18:
    print("You are old enough to vote.")
else:
    print("You are not old enough to vote yet.")
```

```python
# 'if-elif-else' statement
# Use 'elif' (else if) when you have multiple conditions to check.

score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")

score = 55

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

These examples cover the basics of conditional statements. Try changing the values of the variables (`number`, `age`, `score`) and running the cells to see how the output changes! What would you like to explore next?

### Loops (for/while)

Loops allow you to execute a block of code repeatedly. Python offers `for` loops to iterate over sequences and `while` loops to repeat as long as a condition is true.

```python
# 'for' loop
# A 'for' loop is used for iterating over a sequence (that is, a list, tuple, dictionary, set, or string).

# Example 1: Iterating through a list of numbers
numbers = [1, 2, 3, 4, 5]
print("Numbers in the list:")
for num in numbers:
    print(num)

# Example 2: Iterating through a string
word = "Python"
print("\nCharacters in the word:")
for char in word:
    print(char)

# Example 3: Using range() to loop a specific number of times
print("\nCounting from 0 to 4:")
for i in range(5):
    print(i)
```

```python
# 'while' loop
# A 'while' loop executes a set of statements as long as a condition is true.

# Example 1: Counting up to a number
count = 0
print("Counting up to 3:")
while count < 3:
    print(count)
    count += 1 # Increment count to eventually stop the loop

# Example 2: Summing numbers until a limit
current_sum = 0
i = 1
print("\nSumming numbers (1+2+3+...):")
while current_sum < 10:
    current_sum += i
    print(f"Adding {i}, current sum: {current_sum}")
    i += 1
```
